```python
dnkn1 = dnkn.copy()
dnkn1
```

```python
food_items = ['Donuts', 'Sandwiches', 'Muffins', 'Hash Browns', 'Dunkin Bowls', 'Kolache']
dnkn2=dnkn1[~dnkn1['Category'].isin(food_items)].copy()
```

```python
dnkn2.duplicated().sum()
```

```python
dnkn2['Cholesterol (mg)'] = (
    dnkn2['Cholesterol (mg)']
        .replace('<5', '5')
)
```

```python
for col in dnkn2.select_dtypes(include=['object', 'str']).columns:
    dnkn2[col] = (
        dnkn2[col]
            .str.replace('\xa0', '', regex=False)
            .str.strip()
    )
```

```python
for col in dnkn2.select_dtypes(include=['object','str']).columns:
        print('\n', col)
        print(dnkn2[col].unique()[:50])
```

```python
import numpy as np
dnkn2 = dnkn2.replace('',np.nan)
```

```python
dnkn2.isnull().sum()
```

```python
dnkn3 = dnkn2.dropna().copy()
```

```python
dnkn3.isnull().sum()
```

```python
numeric_columns = [
    'Calories',
    'Total Fat (g)',
    'Saturated Fat (g)',
    'Trans Fat (g)',
    'Cholesterol (mg)',
    'Sodium (mg)',
    'Carbs (g)',
    'Fiber (g)',
    'Sugars (g)',
    'Protein (g)',
    'Weight Watcher Pnts'
]
for col in numeric_columns:
    dnkn3[col] = pd.to_numeric(dnkn3[col])
```

```python
print("========== MISSING VALUES ==========")
print(dnkn3.isnull().sum())

print("\n========== DATA TYPES ==========")
print(dnkn3.dtypes)

print("\n========== DUPLICATE RECORDS ==========")
print("Duplicates:", dnkn3.duplicated().sum())

print("\n========== CATEGORY CHECK ==========")
print(dnkn3['Category'].value_counts())

print("\n========== SPECIAL VALUE CHECK ==========")
print("Cholesterol Values:")
print(dnkn3['Cholesterol (mg)'].unique())

print("\nTotal Fat Values:")
print(dnkn3['Total Fat (g)'].unique())

print("\n========== DATASET SHAPE ==========")
print(dnkn3.shape)

print("\n========== SUMMARY STATISTICS ==========")
print(dnkn3.describe().round(2))
```

```python
dnkn3.to_csv(
    'dnkn_clean.csv',
    index=False
)
```
